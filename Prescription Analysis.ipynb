```python
import pandas as pd
name = ["Rohith","Anudeep","Ravi","Bhanu","Charan","Sasi"]
age = [20,21,19,21,22,23]
condition = ["Normal","Moderate","Serious","Abnormal","Normal"]
oxygen_level = [95,92,79,98,93,100]
available_beds = 5
 
for i in range(len(name)):
    if oxygen_level[i] <= 90:
        print(name[i],"- needs icu")
    elif oxygen_level[i] in range(90,94):
        print(name[i],"-need General Ward")
    else:
        print(name[i],"- no observation")
```

```python
for i in range(len(name)):
    if oxygen_level[i] <= 90 and available_beds > 0:
        print(name[i],"- need icu bed")
        available_beds -= 1
    elif oxygen_level[i] in range(91,94) and available_beds > 0:
        print(name[i],"- allocates into general ward beds")
    else:
        print(name[i],"- no need to allocate beds")
        
```

```python
import matplotlib.pyplot as plt

plt.bar(name, oxygen_level)
plt.xlabel("Patients")
plt.ylabel("Oxygen Level")
plt.title("Patient Oxygen Levels")
plt.xticks(rotation=67)
plt.show()
```
