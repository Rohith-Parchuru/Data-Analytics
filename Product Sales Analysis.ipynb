```python
import pandas as pd

data = {
    "Daily Sales": [25, 15, 30, 10, 20, 18, 35],
    "Product Stock": ["Laptop","Smartphone","Headphones","Smart Watch","Bluetooth Speaker","Power Bank","Tablet"],
    "Expiry Date": [30, 25, 15, 20, 10, 18, 28],
    "Seasonal Demand": [ "High","High","Medium","Medium","Low","Medium","High"]
}

df = pd.DataFrame(data)

print(df)
```

```python
avg_sales = df["Daily Sales"].mean()
print("Daily Average Sales :",avg_sales)

for i in range(len(df)):
    if df["Daily Sales"][i] > avg_sales:
        print(df["Product Stock"][i], "is a fast-moving product")
```

```python
for i in range(len(df)):
    if df["Daily Sales"][i] < avg_sales:
        print(df["Product Stock"][i],"is a Least-moving product")
df['Stock Quantity'] = [20, 10, 15, 5, 8, 12, 30]        
```

```python
print("Out of Stock Products are :")
for i in range(len(df)):
    if df["Stock Quantity"][i] < df["Daily Sales"][i]: 
        print(df["Product Stock"][i])
```

```python
import matplotlib.pyplot as plt
demand_count = df["Seasonal Demand"].value_counts()

plt.figure(figsize=(4,4))
plt.pie(
    demand_count,
    labels=demand_count.index,
    autopct="%1.1f%%"
)

plt.title("Seasonal Demand Distribution")
plt.show()
```

```python
out_of_stock = df[df["Stock Quantity"] < df["Daily Sales"]]

plt.figure(figsize=(8,5))
plt.bar(out_of_stock["Product Stock"],
        out_of_stock["Daily Sales"])

plt.title("Products at Risk of Stockout")
plt.xlabel("Products")
plt.ylabel("Daily Sales")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Product Stock"], df["Stock Quantity"])

plt.title("Stock Quantity by Product")
plt.xlabel("Products")
plt.ylabel("Stock Quantity")
plt.xticks(rotation=45)

plt.show()
```
