```python
import pandas as pd
import matplotlib.pyplot as plt

name = ["Rohith","Anudeep","Ravi","Bhanu","Charan","Sasi"]
age = [20,21,19,21,22,23]

condition = ["Normal","Moderate","Serious","Abnormal","Normal","Critical"]

oxygen_level = [95,92,79,98,93,100]

available_beds = 5

bed_allocation = []

# 1. Diagnosis
for i in range(len(name)):
    if oxygen_level[i] <= 90:
        print(name[i], "- needs ICU")
    elif oxygen_level[i] in range(90,94):
        print(name[i], "- need General Ward")
    else:
        print(name[i], "- no observation")


# 2. Bed Allocation (IMPORTANT FIX)
for i in range(len(name)):
    
    if oxygen_level[i] <= 90 and available_beds > 0:
        print(name[i], "- ICU bed allocated")
        bed_allocation.append("ICU Bed Allocated")
        available_beds -= 1
        
    elif oxygen_level[i] in range(91,94) and available_beds > 0:
        print(name[i], "- General Ward bed allocated")
        bed_allocation.append("General Ward Bed")
        available_beds -= 1
        
    else:
        print(name[i], "- No bed allocated")
        bed_allocation.append("No Bed Allocation")


# 3. Visualization - Bar Chart
colors = ["red" if x <= 90 else "green" for x in oxygen_level]

plt.bar(name, oxygen_level, color=colors)
plt.xlabel("Patients")
plt.ylabel("Oxygen Level")
plt.title("Risk Level Visualization")
plt.xticks(rotation=45)
plt.show()


# 4. Pie Chart
icu = bed_allocation.count("ICU Bed Allocated")
general = bed_allocation.count("General Ward Bed")
none = bed_allocation.count("No Bed Allocation")

labels = ["ICU", "General Ward", "No Bed"]
sizes = [icu, general, none]

plt.pie(sizes, labels=labels, autopct="%1.1f%%")
plt.title("Bed Allocation Distribution")
plt.show()
```
