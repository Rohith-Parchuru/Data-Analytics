```python
import pandas as pd

import matplotlib.pyplot as plt

data = {
    "CustomerName":["Rahul","Sneha","Arjun","Priya"],
    "InternetSpeed":[100,50,200,25],
    "PaymentFailure":["Yes","No","No","Yes"],
    "DeliveryCharges":[50,30,20,60],
    "PurchaseCompleted":["No","Yes","Yes","No"]
}

df = pd.DataFrame(data)

print(df)
df.to_csv("e-commerce-dataset",index=False)
abandonment = df[df["PurchaseCompleted"] == "No"]

print("Total Abandonment:", len(abandonment))


print()

Internet = df[df['InternetSpeed'] <= 45]
print("Internet Diagnostics :",Internet)

payment_Diagnostic = df[df['PaymentFailure'] == "No"]
print("Total_PaymentFailures :",payment_Diagnostic.sum())

delivery = df[df['DeliveryCharges'] < 30]
print(delivery.sum(), "delivery will be delay")


print("Root Cause Analysis")

print("Low Internet Speed:",Internet["InternetSpeed"].mean())

print("Payment Failures:",len(payment_Diagnostic))

print("High Delivery Charges:",delivery["DeliveryCharges"].mean())

plt.bar(df["CustomerName"], df["DeliveryCharges"])

plt.title("Payment Failures")

plt.xlabel("Customer Name")
plt.ylabel("Delivery Charges")

plt.show()
```
