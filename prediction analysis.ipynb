```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={
        'Day':[1,2,3,4,5,6,7],
        'Water_liters':[1000,3000,2500,500,1050,3500,1500]
     }
df=pd.DataFrame(data)
print(df)
```

```python
#Average water consumption
avg_water = df['Water_liters'].mean()
print("Average Water Consumption:", avg_water, "liters")
```

```python
#Daily wtaer usage
print("Daily Water Usage:")
print(df[['Day', 'Water_liters']])
```

```python
#water usage increased
df['Usage_Increase'] = df['Water_liters'].diff()

print(df)
```

```python
#sum of daily wyaer increased
total_water = df['Water_liters'].sum()
print("Total Water Consumption:", total_water, "liters")
```

```python
#prediction
from sklearn.linear_model import LinearRegression
x=df[['Day']]
y=df['Water_liters']
model=LinearRegression()
model.fit(x,y)
```

```python
import pandas as pd

future_days = pd.DataFrame({'Day': [8, 9, 10]})

predictions = model.predict(future_days)

for day, pred in zip(future_days['Day'], predictions):
    print(f"Predicted water consumption for Day {day}: {pred:.2f} liters")
```

```python
import matplotlib.pyplot as plt

# Original data
plt.scatter(df['Day'], df['Water_liters'], label="Actual")

# Prediction line
plt.plot(df['Day'], model.predict(x), label="Regression Line")

# Future points
plt.scatter(future_days, predictions, label="Predicted", marker='x')

plt.xlabel("Days")
plt.ylabel("Water Consumption (Liters)")
plt.title("Smart Water Tank Prediction")
plt.legend()
plt.show()
```

```python
plt.pie(df['Water_liters'],
        labels=df['Day'],
        autopct='%1.1f%%',
        shadow=True,
        startangle=90)

plt.axis('equal')  # perfect circle
plt.title("Smart Water Tank - Usage Distribution")
plt.show()
```
