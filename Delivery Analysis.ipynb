```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Order Time": [7, 9, 4, 11, 12,3, 5, 7, 6, 10],
    "Delivery Time": [ 25, 30, 20, 35, 28,22, 40, 18, 27, 32],
    "Delivery Partner": ["Ravi", "Anil", "Ravi", "Suresh", "Anil","Kiran", "Ravi", "Suresh", "Kiran", "Anil"],
    "Customer Ratings": [5, 4, 5, 3, 4,5, 2, 5, 4, 3]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
Avg_DTime = df['Delivery Time'].mean()
print("Average Delivery Time:", Avg_DTime)

Peak_Order_hours = df['Order Time'].max()
print("Peak Order Hour:", Peak_Order_hours)

delay_areas = df['Delivery Time'] - df['Order Time']
print("Delivery Delay:")
print(delay_areas)

for i in range(len(df)):
    if df['Customer Ratings'][i] == 5:
        print(df['Delivery Partner'][i])
```

```python


plt.figure(figsize=(8,5))
plt.bar(df["Delivery Partner"], df["Delivery Time"])

plt.title("Delivery Time by Delivery Partner")
plt.xlabel("Delivery Partner")
plt.ylabel("Delivery Time (minutes)")

plt.show()
```

```python
partner_count = df["Delivery Partner"].value_counts()

plt.figure(figsize=(6,4))
plt.pie(
    partner_count,
    labels=partner_count.index,
    autopct="%1.1f%%"
)

plt.title("Orders Distribution by Delivery Partner")

plt.show()
```
