```python
import pandas as pd
```

```python
data = {
    "Product Purchases": ["Mobile","shirt","Mobile","ear buds","speakers","stove"],
    "Cart Abandonment": [15, 22, 10, 18, 8, 20],
    "Customer Ratings": [4.5, 3.8, 4.9, 4.2, 5.0, 3.9],
    "Payment Method": ["UPI","Credit Card","Debit Card","Cash on Delivery","UPI","Credit Card"]
}

df = pd.DataFrame(data)
df
```

```python
sold_products = df['Product Purchases'].mode()
print("Most Sold Products :", sold_products)

payment_method = df['Payment Method'].mode()
print("Most used payment method :", payment_method)

poor_rating = df[df['Customer Ratings'] < 4 ]

print(poor_rating[['Customer Ratings', 'Product Purchases']])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.bar(df['Product Purchases'], df['Customer Ratings'])

plt.title("Product Purchases vs Customer Ratings")
plt.xlabel("Product Purchases")
plt.ylabel("Customer Ratings")

plt.show()

product_counts = df["Product Purchases"].value_counts()

plt.figure(figsize=(4,6))
plt.pie(product_counts,
        labels=product_counts.index,
        autopct='%1.1f%%')

plt.title("Product Purchase Distribution")
plt.show()
```
